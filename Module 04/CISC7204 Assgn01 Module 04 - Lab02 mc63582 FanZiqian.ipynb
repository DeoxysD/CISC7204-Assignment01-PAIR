This function will download the dataset into your browser


````python
# 本机没有 pyodide，改用 aiohttp 下载清洗后的笔记本价格数据
import aiohttp

async def download(url, filename):
    async with aiohttp.ClientSession() as session:
        async with session.get(url) as response:
            if response.status == 200:
                data = await response.read()
                with open(filename, "wb") as f:
                    f.write(data)
````

# Task 1 : Single Linear Regression

You have learnt that "CPU_frequency" is the parameter with the lowest p-value among the different features of the dataset. Create a single feature Linear Regression model that fits the pair of "CPU_frequency" and "Price" to find the model for prediction.


````python
# Write your code below and press Shift+Enter to execute
# Task 1：简单线性回归——只用 CPU_frequency 预测 Price
# EDA 里它的 P 值最低，所以先拿单变量练手
# X 必须是二维（DataFrame）；Y 是一维价格
lm = LinearRegression()
X = df[["CPU_frequency"]]
Y = df["Price"]
lm.fit(X, Y)
Yhat = lm.predict(X)  # 用训练数据做样本内预测，后面用来画图、算误差
 
````

Evaluate the Mean Squared Error and R^2 score values for the model. 


````python
# Write your code below and press Shift+Enter to execute
# R²：模型解释了多少价格波动（越接近 1 越好）
# MSE：预测偏差平方的平均（越小越好）
mse_slr = mean_squared_error(df["Price"], Yhat)
r2_score_slr = lm.score(X, Y)
print("The R-square for Linear Regression is: ", r2_score_slr)
print("The mean square error of price and predicted value is: ", mse_slr)
 
````

# Task 3 - Polynomial Regression
Use the variable "CPU_frequency" to create Polynomial features. Try this for 3 different values of polynomial degrees. Remember that polynomial fits are done using `numpy.polyfit`. 


````python
#  Write your code below and press Shift+Enter to execute
# Task 3：只用 CPU_frequency，分别拟合 1、3、5 阶多项式
# polyfit 需要一维 x，所以把 DataFrame 压成数组（不要用二维的 X）
X = np.asarray(df["CPU_frequency"]).flatten()

# 1 阶等价于直线；3、5 阶可以弯曲，更贴点，但也更容易过拟合
f1 = np.polyfit(X, Y, 1)
p1 = np.poly1d(f1)

f3 = np.polyfit(X, Y, 3)
p3 = np.poly1d(f3)

f5 = np.polyfit(X, Y, 5)
p5 = np.poly1d(f5)

````

Plot the regression output against the actual data points to note how the data fits in each case. To plot the polynomial response over the actual data points, you have the function shown below.


````python
def PlotPolly(model, independent_variable, dependent_variabble, Name):
    x_new = np.linspace(independent_variable.min(),independent_variable.max(),100)
    y_new = model(x_new)

    plt.plot(independent_variable, dependent_variabble, '.', x_new, y_new, '-')
    plt.title(f'Polynomial Fit for Price ~ {Name}')
    ax = plt.gca()
    ax.set_facecolor((0.898, 0.898, 0.898))
    fig = plt.gcf()
    plt.xlabel(Name)
    plt.ylabel('Price of laptops')
````

Also, calculate the R^2 and MSE values for these fits. For polynomial functions, the function sklearn.metrics.r2_score will be used to calculate R^2 values.


````python
#  Write your code below and press Shift+Enter to execute
# 多项式模型不是 sklearn 的 LinearRegression，要用 r2_score(真实值, 预测值)
# p1(X) 表示把 CPU_frequency 代入 1 阶多项式得到预测价格
r_squared_1 = r2_score(Y, p1(X))
print("The R-square value for 1st degree polynomial is: ", r_squared_1)
print("The MSE value for 1st degree polynomial is: ", mean_squared_error(Y, p1(X)))
r_squared_3 = r2_score(Y, p3(X))
print("The R-square value for 3rd degree polynomial is: ", r_squared_3)
print("The MSE value for 3rd degree polynomial is: ", mean_squared_error(Y, p3(X)))
r_squared_5 = r2_score(Y, p5(X))
print("The R-square value for 5th degree polynomial is: ", r_squared_5)
print("The MSE value for 5th degree polynomial is: ", mean_squared_error(Y, p5(X)))

````

# Task 4 - Pipeline
Create a pipeline that performs parameter scaling, Polynomial Feature generation and Linear regression. Use the set of multiple features as before to create this pipeline. 


````python
#  Write your code below and press Shift+Enter to execute
# Task 4：Pipeline 按顺序执行：标准化 → 多项式特征 → 线性回归
# include_bias=False：不额外生成全 1 列（线性回归自己有截距）
# 特征仍用 Task 2 的 Z（多个配置变量）
Input = [
    ("scale", StandardScaler()),
    ("polynomial", PolynomialFeatures(include_bias=False)),
    ("model", LinearRegression()),
]
pipe = Pipeline(Input)
Z = Z.astype(float)  # StandardScaler 需要浮点数
pipe.fit(Z, Y)
ypipe = pipe.predict(Z)

````

Evaluate the MSE and R^2 values for the this predicted output.


````python
#  Write your code below and press Shift+Enter to execute
# 多元 + 多项式特征后，样本内 R² 通常比单纯多元线性更高、MSE 更低
# 注意：这是在全部数据上评估，后面还要用测试集才能判断有没有过拟合
print("MSE for multi-variable polynomial pipeline is: ", mean_squared_error(Y, ypipe))
print("R^2 for multi-variable polynomial pipeline is: ", r2_score(Y, ypipe))

````
